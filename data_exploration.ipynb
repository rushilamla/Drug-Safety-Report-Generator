```python
import pandas as pd

def count_items(value):
    if pd.isna(value):
        return 0
    return len(str(value).split(","))

df["reaction_count"] = df["patient_reaction_reactionmeddrapt"].apply(count_items)
df["outcome_count"] = df["patient_reaction_reactionoutcome"].apply(count_items)

print("Reaction count distribution:")
print(df["reaction_count"].value_counts().sort_index())

print("\nOutcome count distribution:")
print(df["outcome_count"].value_counts().sort_index())

print("\nRows where counts don't match:")
mismatch = df[df["reaction_count"] != df["outcome_count"]]

print("Number of mismatched rows:", len(mismatch))
```

```python
mismatch = df[df["reaction_count"] != df["outcome_count"]]

print(
    mismatch[
        [
            "safetyreportid",
            "patient_reaction_reactionmeddrapt",
            "patient_reaction_reactionoutcome"
        ]
    ].to_string(index=False)
)
```

```python
df["age_years"] = pd.to_numeric(df["age_years"], errors="coerce")
```

```python
bins = [-float("inf"), 18, 45, 65, 75, float("inf")]
labels = ["<18", "18–44", "45–64", "65–74", "75+"]

df["age_group"] = pd.cut(
    df["age_years"],
    bins=bins,
    labels=labels,
    right=False
)
```
